>> # BANKING MANAGEMENT SYSTEM USING JAVA OOPS CONCEPTS
        
## PROBLEM STATEMENT: 
    To build a Simple Banking Management System using OOPs Concepts available in Java.
    It follows the UML diagram provided in the document with entities like Account, Bank, Branch, Customer and Transaction. 

## FEATURES:
> - Create a Customer <br>
> - Create Savings/Current Account for a Customer <br>
> - Deposit money into an Account <br>
> - Withdraw money from an Account <br>
> - Transfer money between two Accounts <br>
> - Freeze/Unfreeze an Account <br>
> - View Account details and Transaction history <br>
    
## EXAMPLE USAGE: 
> - Create a Bank
> - Add Branches to the Bank
> - Create a Clerk and Manager for each Branch
> - Clerk creates Customers
> - Manager creates Accounts for Customers
> - Perform transactions like deposit, withdraw, transfer, freeze/unfreeze
> - View Account details and Transaction history
> - Handle exceptions for invalid operations
> -  Display appropriate messages for each operation

## Following are classes and interfaces in each packages.
### 1. package interfaces
   - **interface Freezable**                     --handles freeze and unfreeze for an account.
   - **interface Transactable**                  --makes transactions.
   - **interface AccountViewer**                 --declares methods to view account details such as account number, balance, type, frozen status, and transaction history. It extends Freezable and Transactable.

### 2. package bankexceptions
   - **class AccountNotFoundException**            --for Account is not present
   - **class AccountFrozenException**              --for Transaction made in Frozen Account
   - **class CustomerNotFoundException**           --for Customer is not Found
   - **class InsufficientFundsException**          --if user wants to withdraw beyond the balance
   - **class InvalidAmountException**              --if user enters invalid value for withdraw or deposit amount (say, negative number).

     -- All the above class extends Exception class.

### 3. package account
- **abstract class Account**  --Abstract class that defines common fields (accountNumber, balance, customer, transactionList, frozen status) and abstract methods for deposit, withdraw, transfer
- **class CurrentAccount**    --Current account without interest calculation
- **class SavingsAccount**    --Savings account with interest calculation
- **class Transaction**       --To hold the details of each transaction. 

### 4. package User
- **class Customer**          --To hold customer details like name, age, address, phone number and list of accounts.
- **class CustomerConsole**   --Runs the console for Customer dashboard

### 5. package bank
- **class Clerk**             --For one who creates a customer.
- **class Manager**           --For one who creates an account for a customer.
- **class Branch**            --To hold branch details like branch name, branch code, manager, clerk and list of customers.
- **class Bank**              --To hold bank details like bank name, IFSC code and list of branches.
    
### 6. package database
- **class DataStorage**       --Stores the account, customer and transaction data to files.
- **class FileHandling**      --Handles files

### 6. package database
- **class DataStorage**        --uses methods available in FileHandling class such as, saveTofile for loading the account, customer and transaction data from console to file and loadFromfile to readback from file and store it to Collections like list for searching, modification and creation.
- **class FileHandling**       --Load and Store data to file using Input/OutputStreams
